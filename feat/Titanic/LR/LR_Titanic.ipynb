```python
# Select final features for model training.
features = ['Sex_encoded', 'Pclass_1', 'Pclass_2', 'Pclass_3', 'Age_filled',
            'SibSp', 'Parch', 'Fare']
X = train[features] # all features
Y = train['Survived'] # target variable

print("features after process:")
print(X.head()) # it works, good

# Split the data  

# Split into training, validation, and test sets
X_train_val, X_test, Y_train_val, Y_test = train_test_split(X, Y,
test_size=0.2, random_state=42)

# test_size=0.2 means 20% for the second part
# random_state=42 means a fixed random seed for reproducibility
# from sklearn.model_selection import train_test_split
# test_size: float, int, or None, default=None
# cannot run with test_size=None

X_train, X_val, Y_train, Y_val = train_test_split(X_train_val, Y_train_val,
test_size=0.25, random_state=42)

print(f"train: {X_train.shape[0]} people")
print(f"validation: {X_val.shape[0]} people")
print(f"test: {X_test.shape[0]} people")
# 0 means all rows, 1 means all columns
```

```python
# starting training
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Initialize the Logistic Regression model
model = LogisticRegression(max_iter=1000)
# max_iter means maximum iterations
```

#——————————————————————————————————————————————————————————————————————
# Empirical max_iter guide (for initial use):
# n = number of samples, d = number of features
# Small (n<10k, d<100): 100-300  # near-instant
# Medium (n~100k, d~1k): 1000-2000  # L-BFGS stops early
# Large (n>1M, d>10k): Use 'saga', 500-1000  # SGD
# lbfgs and saga are solvers, solver like SGD means optimization algorithm

```python
# study the training data
history = model.fit(X_train, Y_train)
# fit() means training, fit the model according to the given training data
# to check the learned weights (coefficients)
# to calculate the θ in the formula
print("learned weights (coefficients):")
for feature, coef in zip(features, model.coef_[0]):
   print(f"{feature}: {coef:.3f}") # it is round half to even
print(f"intercept: {model.intercept_[0]:.3f}")
print(model.classes_)
# zip() can combine two lists into a list of tuples    
#——————————————————————————————————————————————————————————————————————
```

# every time we use zip(), it will create a new iterator
# in case we assign zip() to a variable and use it multiple times
# it will be exhausted after the first use and output nothing
# in the second use

# zip() is a iterator, so it can be used only once
# zip() processes data one by one, so it is memory efficient
# if you need to use it multiple times, convert it to a list first
# when the number of two lists is different, zip() stops at the end
# of the shorter one, AI inspired me this case

#——————————————————————————————————————————————————————————————————————
# Iterator: for lazy, one-pass consumption (memory efficient, but exhausted).
# Copy: for safe, independent data (memory cost, but no side effects).
# View: for on-the-fly, linked data (memory efficient, but changes propagate).
#——————————————————————————————————————————————————————————————————————
# zip() returns iterator with internal position tracking in CPython 
# implementation.
# This pointer is maintained at C level, invisible from Python code.
# I never know python has various versions for different languages before.

# KEY INSIGHTS:

# 1. Python as 'Secondary' Language:
#    - Most Python runtimes (CPython) implemented in lower-level languages (C)
#    - This makes Python operate at higher abstraction level

# 2. C Language in Embedded Systems:
#    - C is DOMINANT but NOT exclusive for embedded development
#    - Alternatives: C++, Rust, Assembly for different use cases
#    - C preferred for: direct hardware access, minimal runtime overhead

# 3. Linux Role Clarification:
#    - Linux is Unix-like KERNEL, not programming language
#    - Used in embedded via Embedded Linux distributions
#    - Many embedded Linux systems programmed in C, but Python runs on top
#    - For bare-metal systems without OS, C remains primary choice

```python
# test on validation set
val_predictions = model.predict(X_val)
val_accuracy = accuracy_score(Y_val, val_predictions)
print(f"\nthe validation accuracy: {val_accuracy:.2%}") #
# the weight of female is positive, weight of Pclass_3 is negative
# which is the same as our intuition

# 4. Model evaluation on the test set
test_predictions = model.predict(X_test)
test_accuracy = accuracy_score(Y_test, test_predictions)
print(f"\n!!!final test accuracy:{test_accuracy:.2%} !!!")
```
